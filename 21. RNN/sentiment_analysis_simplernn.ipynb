```python
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index)
```

we have 17 Unique words

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

17 is the size of the vocabulary
2 is in how many numbers you want to represent your input vector

5 represents how many words do you have

In Keras, the Embedding layer's first argument (input_dim) represents the total size of the vocabulary, which must be maximum_integer_index + 1.


In the code:

1. `tokenizer.word_index` has 17 words, so words are assigned integers from 1 to 17.
2. Index 0 is reserved for padding (`pad_sequences`).
3. The valid token indices therefore range from 0 to 17 (a total of 18 distinct values).


```python
# Change input_dim from 17 to len(tokenizer.word_index) + 1 (18)
vocab_size = len(tokenizer.word_index) + 1  # 18
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=2, input_length=5))


model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
# 1. Reload the dataset with num_words=10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
# 2. Pad the sequences to length 50
X_train = pad_sequences(X_train, padding='post', maxlen=50)
X_test = pad_sequences(X_test, padding='post', maxlen=50)
```

```python
X_train.shape
```

```python
# 3. Define and compile the model
model = Sequential()
model.add(Embedding(input_dim=10000, output_dim=2, input_length=50))
model.add(SimpleRNN(32, return_sequences=False))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
```

```python
# 4. Train the model
history = model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```
