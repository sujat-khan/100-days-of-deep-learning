```python
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img
from tensorflow.keras.preprocessing import image
```

```python
# Load and resize the image
img = load_img('Train/cat.jpeg', target_size=(200, 200))
```

```python
# Display the image
plt.imshow(img)
plt.axis('off')  # Optional: hides axis coordinates
plt.show()
```

```python
type(img)
```

```python
# Instantiating the generator configuration
datagen = ImageDataGenerator(
    #rescale=1./255,            # Normalize pixel values from [0, 255] to [0, 1]
    rotation_range=40,         # Randomly rotate images up to 40 degrees
    width_shift_range=0.2,     # Randomly shift images horizontally (20% of width)
    height_shift_range=0.2,    # Randomly shift images vertically (20% of height)
    shear_range=0.2,           # Shear transformations
    zoom_range=0.2,            # Randomly zoom inside images
    horizontal_flip=True,      # Randomly flip images horizontally
    #fill_mode='nearest'        # Strategy used to fill in newly created pixels, can be constant as well
)
```

```python
img = image.img_to_array(img)
```

```python
type(img)
```

```python
img.shape
```

```python
input_batch = img.reshape(1,200,200,3)
```

```python
i = 0

for output in datagen.flow(input_batch, batch_size=1, save_to_dir='aug'):
    i = i + 1
    
    if i == 10:
        break
```
